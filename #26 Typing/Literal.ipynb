### 🔹 What is  `Literal`  in Python?

`Literal`  is a type hint introduced in  **PEP 586**  (Python 3.8+) that allows you to specify that a variable or function parameter must be exactly one of a set of specific values.


### 🔹 Why use  `Literal`?

It helps with:

-   **Type safety**: Ensures only specific values are allowed.
-   **Better autocomplete**  in IDEs.
-   **Static type checking**  with tools like  `mypy`.

```python
from typing import Literal

def get_status(code: Literal[200, 404, 500]) -> str:
    if code == 200:
        return "OK"
    elif code == 404:
        return "Not Found"
    elif code == 500:
        return "Server Error"

# Valid call
print(get_status(200))  # OK
```

```python
print(get_status(500))
```

```python
print(get_status(404))
```

```python
# Invalid call (type checker will flag this)
print(get_status(403))  # ❌ Not allowed
```

```python

from typing import Dict, List, Literal, cast

Status = Literal["success", "error", "pending"]

def handle_response(status: Status) -> str:
    return f"Status is {status}"

# Using cast to assure type checker
response = cast(Status, "success")
print(handle_response(response))
```

```python
# Using cast to assure type checker
response = cast(Status, "failure")
print(handle_response(response))
```
