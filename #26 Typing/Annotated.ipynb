 In Python, `Annotated` is a feature from the `typing` module that allows you to **attach metadata** to a type hint
 
This is useful when you want to provide **additional information** about a variable, parameter, or return type — such as validation rules, documentation hints, or framework-specific annotations — without changing the actual type.

```python

from typing import Annotated

def greet(name: Annotated[str, "User's full name"]) -> None:
    print(f"Hello, {name}!")
```

Here, "User's full name" is metadata that doesn't affect runtime behavior but can be used by tools or frameworks.
