
A **TypedDict** in Python is a feature from the `typing` module that allows you to define dictionaries with a specific structure — meaning you can specify what keys are expected and what types their values should be. This is especially useful for **type checking** and **code clarity**.

### Why use  `TypedDict`?

-   Helps with  **static type checking**  using tools like  `mypy`  or  `pyright`.
-   Makes your code  **self-documenting**.
-   Reduces bugs by ensuring consistent dictionary structure.

```python
from typing import TypedDict

class Employee(TypedDict):
    name: str
    age: int
    department: str

def print_employee_info(emp: Employee) -> None:
    print(f"{emp['name']} is {emp['age']} years old and works in {emp['department']}.")

# Usage
employee_data = {
    "name": "Santhosh",
    "age": 30,
    "department": "Engineering"
}

print_employee_info(employee_data)
```
