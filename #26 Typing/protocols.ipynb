**`Protocol`  in Python is similar to interfaces** in other languages like Java or C#. It defines a set of methods and properties that a class must implement to be considered compatible, **without requiring inheritance**.

### What is  `Protocol`?

A  `Protocol`  allows you to define an interface that other classes can implement without explicitly inheriting from it. If a class has the required methods and properties defined in the protocol, it is considered a subtype of that protocol.


```python
from typing import Protocol

class Greeter(Protocol):
    def greet(self, name: str) -> str:
        ...

class FriendlyGreeter:
    def greet(self, name: str) -> str:
        return f"Hello, {name}!"

def welcome(greeter: Greeter, name: str) -> None:
    print(greeter.greet(name))

# This works because FriendlyGreeter matches the Greeter protocol
friendly = FriendlyGreeter()
welcome(friendly, "Santhosh")
```
