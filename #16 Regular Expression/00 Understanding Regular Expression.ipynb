```python
# Metacharacters

#1. []                           A set of Character                                        [a-m]
#2. \                            Signals a special sequence                                 \d
#3. .                            Any character                                             he..o
#4. ^                            starts with                                               ^hello
#5. $                            Ends with                                                 bye$
#6. *                            Zero or more occurences                                   aix*
#7. +                            One or more occurences                                    aix+
#8. {}                           Exactly the specified number of occurences                al{2}
#9. |                            Either or                                                 flass|stays
#10.Capture and group            
```

```python
# Special Sequences

# \A          Return match mentioned begining of the string                                               \AThe
# \b          Return match mentioned beginning or at the end of a word                                    \bThe or The\b
# \B          Return match mentioned, but NOT at the beginning (or at the end) of a word                  \Bain
# \d          Return match where the string contains digits                                               \d
# \D          Return match where the string does not contains digits                                      \D
# \s          Returns a match where the string contains a white space character                           \s
# \S          Returns a match where the string DOES NOT contain a white space character                   \S
# \w          Returns a match where the string contains any word characters                               \w
# \W          Returns a match where the string DOES NOT contain any word characters                       \W
# \Z          Returns a match if the specified characters are at the end of the string                    Spain\Z
```

```python
# Sets 

# [arn]       Returns a match where one of the specified characters (a, r, or n) are present
# [a-n]       Returns a match for any lower case character, alphabetically between a and n
# [^arn]      Returns a match for any character EXCEPT a, r, and n
# [0123]      Returns a match where any of the specified digits (0, 1, 2, or 3) are present
# [0-9]       Returns a match for any digit between 0 and 9
# [0-5][0-9]  Returns a match for any two-digit numbers from 00 and 59
# [a-zA-Z]    Returns a match for any character alphabetically between a and z, lower case OR upper case
# [+]         
```

### Anchors: ^ $ \A \b \Z

```python
'''

1. ^Python : Match "Python" at the start of a string or internal line

2. Python$ : Match "Python" at the end of a string or line

2. ^The end$  :  exact string match (starts and ends with The end)

3. \APython : Match "Python" at the start of a string

4. Python\Z : Match "Python" at the end of a string

5. \bPython\b : Match "Python" at a word boundary

6. \brub\B : \B is nonword boundary: match "rub" in "rube" and "ruby" but not alone

7. Python(?=!) : Match "Python", if followed by an exclamation point.

8. Python(?!!) : Match "Python", if not followed by an exclamation point.

9. \< : Start of word

10. \> : End of word
```

### Quantifiers — * + ? and {}

```python
'''
abc* : Matches a string that has ab followed by zero or more c

abc+ : Matches a string that has ab followed by one or more c

abc{2} : Matches a string that has ab followed by 2 c

abc{2,} : matches a string that has ab followed by 2 or more c

abc{2,5} : matches a string that has ab followed by 2 up to 5 c

a(bc)* : matches a string that has a followed by zero or more copies of the sequence bc

a(bc){2,5} : matches a string that has a followed by 2 up to 5 copies of the sequence bc

\d{3} : Match exactly 3 digits

\d{3,} : Match 3 or more digits

\d{3,5} : Match 3, 4, or 5 digits
```

### OR operator — | or []

```python
'''
a(b|c)  : matches a string that has a followed by b or c

a[bc]   : same as previous, but without capturing b or c
```

### Character Classes:

```python
'''
. :  Matches any character

\d : Matches a single character that is a digit: [0-9]

\w :  matches a word character (alphanumeric character plus underscore) [A-Za-z0-9_]

\s : matches a whitespace character (includes tabs and line breaks) : [ \t\r\n\f]

\D :  Matches a single non-digit character: [^0-9]

\S : Match nonwhitespace: [^ \t\r\n\f]

\W :Match a nonword character: [^A-Za-z0-9_]

\$\d : Matches a string that has a $ before one digit 
```

### Grouping and capturing — ()

```python
'''
a(bc)   : parentheses create a capturing group with value bc

a(?:bc)* : using ?: we disable the capturing group

a(?<foo>bc) : using ?<foo> we put a name to the group
```

### Sets :

```python
'''

[abc]   : Matches a string that has either an a or a b or a c

[aeiou] :Match any one lowercase 'vowel'

[0-9] : Match any 'digit'; same as '[0123456789]'

[a-z] : Match any 'lowercase ASCII letter'

[A-Z] : Match any 'uppercase ASCII letter'

[a-zA-Z0-9]: Match any of the above ('lowercase ASCII letter' and 'uppercase ASCII letter' and 'digit')

[^abc] : Any character _except_ a, b, or c (negation)

[^aeiou] : Match anything other than a 'lowercase vowel'

[^0-9] : Match anything other than a 'digit'

[0-9]%  : A string that has a character from 0 to 9 before a % sign

[^a-zA-Z]  : A string that has not a letter from a to z or from A to Z.
```

### Characters

```python
Escape character: \
Any character: .
Digit: \d
Not a digit: \D
Word character: \w
Not a word character: \W
Whitespace: \s
Not whitespace: \S
Word boundary: \b
Not a word boundary: \B
Beginning of a string: ^
End of a string: $
```

### Groupings

```python
Matches characters in brackets: [ ]
Matches characters not in brackets: [^ ]
Either or: |
Capturing group: ( )
```

### Quantifiers

```python
0 or more: *
1 or more: +
0 or 1: ?
An exact number of characters: { }
Range of number of characters: {Minimum, Maximum}
```

```python
regex = r"([a-zA-Z]+) (\d+)" # text with a-z&A-Z followed by digits
 
match = re.search(regex, "I was born on June 24")

# So this will print "June 24"
print("Full match: %s" % (match.group(0)))

# So this will print "June"
print("Month: %s" % (match.group(1)))

 # So this will print "24"
print ("Day: %s" % (match.group(2)))
```

### Sub

```python
txt = "The rain in Spain"
x = re.sub("\s","@",txt,2)  # replace first 2 characters

print(x)
```

```python
# ^ : Matches beginning of line

re.sub("^The","@","The banglore is The great city") # replace 'The' with '@' only found in the beginning of the sentence
```

```python
re.sub("city$","@","The banglore is The great city")  # replace 'city' with '@' only at the end of the line
```

```python
re.sub("[abd]","@","santhosh b")  # replace any character find on  the [abd] to "@" 
```

```python
re.sub("[^abd]","@","santhosh b") # replace any character not find on  the [abd] to "@" 
```

```python
import re

# Print a list of all matches:
str = "The rain in Spain"
x = re.findall("ai", str)
print(x)

# Return an empty list if no match was found:
x = re.findall("Portugal", str)
print(x)

# Search
x = re.search("\s", str)
print("The first white-space character is located in position:", x.start())

# Split
x = re.split("\s", str)
print(x)

# Split the string only at the first occurrence:
x = re.split("\s", str, 1)
print(x)

# Replace every white-space character with the number 9:
x = re.sub("\s", "9", str)
print(x)

# Replace the first 2 occurrences
x = re.sub("\s", "9", str, 2)
print(x)
```

```python
# Sample code to remove a regex pattern 
import re
text = "remove this #hashtag from analytics vidhya"
regex_pattern = "#[\w]*"

urls = re.finditer(regex_pattern,text)

for i in urls:
    inputtext = re.sub(i.group().strip(),'',text)
print(inputtext)
```
