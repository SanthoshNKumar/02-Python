
In **Pydantic**, a `Field` is used to provide **extra metadata**, **validation rules**, and **default values** for model attributes. It enhances how Pydantic handles data parsing and validation.

```python
from pydantic import BaseModel, Field

class User(BaseModel):
    name: str = Field(..., description="The user's full name")
    age: int = Field(..., gt=0, lt=150, description="Age must be between 1 and 149")
```

-   `...`  means the field is  **required**.
-   `gt=0`  means  **greater than 0**.
-   `lt=150`  means  **less than 150**.
-   `description`  is useful for documentation (e.g., FastAPI).

```python

class Product(BaseModel):
    name: str = Field(..., min_length=3, max_length=50)
    price: float = Field(..., gt=0)
    in_stock: bool = Field(default=True)
```

```python
from pydantic import BaseModel, Field
from typing import Optional

class Product(BaseModel):
    # Required field with a title and description
    name: str = Field(..., title="Product Name", description="The name of the product")

    # Optional field with a default value
    in_stock: bool = Field(default=True, description="Is the product available in stock")

    # Field with numeric constraints
    price: float = Field(..., gt=0, lt=10000, description="Price must be > 0 and < 10,000")

    # Field with string length constraints
    sku: str = Field(..., min_length=5, max_length=10, description="Stock Keeping Unit code")

    # Field with regex validation
    category: str = Field(..., pattern="^[A-Za-z ]+$", description="Category must contain only letters and spaces")

    # Field with alias and example
    discount_percentage: Optional[float] = Field(
        default=None,
        alias="discount",
        ge=0,
        le=100,
        description="Discount percentage (0-100)",
        example=15.0
    )
```

### 🔍 What Each Field Does:

| **Field**              | **Parameters Used**                          | **Purpose**                                           |
|------------------------|----------------------------------------------|-------------------------------------------------------|
| `name`                 | `title`, `description`                       | Adds metadata for documentation                       |
| `in_stock`             | `default`, `description`                     | Optional with a default value                         |
| `price`                | `gt`, `lt`, `description`                    | Must be between 0 and 10,000                          |
| `sku`                  | `min_length`, `max_length`, `description`    | Enforces string length                                |
| `category`             | `regex`, `description`                       | Must match a pattern (only letters and spaces)        |
| `discount_percentage`  | `alias`, `ge`, `le`, `example`, `description`| Optional, with alias and range validation             |
